```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df =pd.read_csv("Books_Data_Clean.csv")
```

```python

df.head()
```

```python
df.describe()
```

```python
df = df[df["Publishing Year"] > 1900]
```

```python
df.isna().sum()
```

```python
df.dropna(subset="Book Name", inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.nunique()
```

```python
plt.hist(df["Publishing Year"])
plt.xlabel("Publishing Year")
plt.ylabel("Frequency")
plt.title("Distribution of Publishing Year")
plt.show()
```

```python
df["genre"].value_counts().plot(kind = "bar")
plt.xlabel("Genre")
plt.ylabel("Frequency")
plt.title("Genre Distribution")
plt.show()
```

```python
df.groupby("Author")["Book_average_rating"].mean().sort_values(ascending=False).head()

```

```python
sns.boxplot(x="genre", y="Book_ratings_count", data=df)
plt.xlabel("Genre")
plt.ylabel("Book Rating Count")
plt.title("Boxplot of Book Rating Count for each genre")
plt.show()
```

```python
plt.scatter(df["sale price"], df["units sold"])
plt.xlabel("sale Price")
plt.ylabel("units scale")
plt.title("Scatter plot of Book Average Rating vs Book Rating Count")
plt.show()
```

```python
df["language_code"].value_counts().plot(kind = "bar")
plt.xlabel("Language")
plt.ylabel("Frequency")
plt.title("Language Distribution")
plt.show()
```

```python
language_counts = df["language_code"].value_counts()
plt.pie(language_counts, labels=language_counts.index, startangle=90, autopct="%1f%%")
plt.title("Language Distribution")
plt.show()
```

```python
df.groupby("Publisher ")["publisher revenue"].sum().sort_values(ascending=False)
```

```python
df.groupby("Author_Rating")["Book_ratings_count"].mean().sort_values(ascending=False).max()
```

```python
df.groupby("language_code").size().sort_values(ascending=False)
```

```python
plot =df.groupby("Author_Rating")["Book_ratings_count"].mean()
print(plot)
plt.pie(plot, labels=plot.index, startangle=90, autopct="%1f%%")
plt.show()
```

```python
df.groupby("Author_Rating")["Book_ratings_count"].var()
```

```python
plt.scatter(df["Book_average_rating"], df["Book_ratings_count"])
plt.xlabel("Book Average Rating")
plt.ylabel("Book Rating Count")
plt.title("Scatter plot of Book Average Rating vs Book Rating Count")
plt.show()
```

```python
total_gross_sales_by_author = df.groupby("Author")["gross sales"].sum()
total_gross_sales_by_author.sort_values(ascending=False).head(10).plot(kind="bar")
plt.xlabel("Author")
plt.ylabel("Total Gross Sales")
plt.title("Top 10 Authors by Gross Sales")
plt.show()
```

```python
sns.boxplot(x="Author_Rating", y="units sold", data=df)
plt.xlabel("Author_Rating")
plt.ylabel("units sold")
plt.title("Boxplot of units sold for each Author_Rating")
plt.show()
```

```python
df.groupby("Publishing Year")["units sold"].sum().plot(kind="line", marker="o")
plt.xlabel("Publishing Year")
plt.ylabel("Total Units Sold")
plt.title("Total Units Sold by Publishing Year")
plt.show()
```
